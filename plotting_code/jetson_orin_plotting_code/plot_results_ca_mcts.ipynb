```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import re

def parse_results(text):
    rows = []
    seen = set()
    for line in text.strip().split("\n"):
        parts = line.split()
        if len(parts) < 7 or parts[0] not in ("Sepang", "YasMarina"):
            continue
        # Deduplicate by (track, mu, budget, trial, seed)
        key = (parts[0], parts[1], parts[2], parts[3], parts[4])
        if key in seen:
            continue
        seen.add(key)
        collision = 1 if parts[6] == "YES" else 0
        lap_times = []
        match = re.search(r'\[([^\]]+)\]', line)
        if match and match.group(1).strip():
            lap_times = [float(t.replace('s', '')) for t in match.group(1).split()]
        rows.append({
            "track": parts[0],
            "mu": float(parts[1]),
            "budget": int(parts[2].replace("ms", "")),
            "trial": int(parts[3]),
            "seed": int(parts[4]),
            "collision": collision,
            "lap_times": lap_times,
            "fastest_lap": min(lap_times) if lap_times else np.nan,
            "avg_lap": np.mean(lap_times) if lap_times else np.nan,
        })
    return rows

# --- Parse both results ---
df_orig = pd.DataFrame(parse_results(results))
df_rerun = pd.DataFrame(parse_results(results_remain))

# --- Merge: for experiments that were re-run, replace the original result ---
key_cols = ["track", "mu", "budget", "trial", "seed"]
rerun_keys = set(df_rerun[key_cols].apply(tuple, axis=1))

df_keep = df_orig[~df_orig[key_cols].apply(tuple, axis=1).isin(rerun_keys)]
df = pd.concat([df_keep, df_rerun], ignore_index=True)

print(f"Original: {len(df_orig)} rows ({df_orig['collision'].sum()} collisions)")
print(f"Re-run:   {len(df_rerun)} rows ({df_rerun['collision'].sum()} collisions)")
print(f"Merged:   {len(df)} rows ({df['collision'].sum()} collisions)")
print(f"\nTracks: {sorted(df['track'].unique())}")
print(f"Mus: {sorted(df['mu'].unique())}")
print(f"Budgets: {sorted(df['budget'].unique())}")
```

```python
# --- Crash percentage heatmaps: one per track (x=budget, y=mu) ---
crash_pct = df.groupby(["track", "mu", "budget"])["collision"].mean() * 100
crash_pct = crash_pct.reset_index().rename(columns={"collision": "crash_pct"})

tracks = sorted(df["track"].unique())
mus = sorted(df["mu"].unique())
budgets = sorted(df["budget"].unique())

fig, axes = plt.subplots(1, 2, figsize=(18, 6))

for ax, track in zip(axes, tracks):
    sub = crash_pct[crash_pct["track"] == track]
    pivot = sub.pivot(index="mu", columns="budget", values="crash_pct")
    pivot = pivot.reindex(index=mus, columns=budgets).fillna(0)

    im = ax.imshow(pivot.values, aspect="auto", cmap="RdYlGn_r", vmin=0, vmax=100)

    for i in range(len(mus)):
        for j in range(len(budgets)):
            val = pivot.values[i, j]
            color = "white" if val > 50 else "black"
            ax.text(j, i, f"{val:.0f}%", ha="center", va="center",
                    fontsize=11, fontweight="bold", color=color)

    ax.set_xticks(range(len(budgets)))
    ax.set_xticklabels([f"{b}ms" for b in budgets], rotation=45)
    ax.set_yticks(range(len(mus)))
    ax.set_yticklabels([f"{m:.1f}" for m in mus])
    ax.set_xlabel("Computational budget", fontsize=12)
    ax.set_ylabel("Friction coefficient (μ)", fontsize=12)
    ax.set_title(f"{track}", fontsize=13, fontweight="bold")

fig.suptitle("CA-MCTS (Jetson Orin) — Crash Percentage (10 trials each)", fontsize=14)
fig.subplots_adjust(right=0.88)
cbar_ax = fig.add_axes([0.90, 0.15, 0.02, 0.7])
fig.colorbar(im, cax=cbar_ax, label="Crash %")
# plt.savefig("crash_pct_ca_mcts_jetson.png", dpi=150, bbox_inches="tight")
# plt.savefig("crash_pct_ca_mcts_jetson.pdf", bbox_inches="tight")
plt.show()
```

```python
# --- TikZ heatmap for fastest lap time (full column width, no legend) ---
df_ok = df[df["collision"] == 0]
mus = sorted(df["mu"].unique())
budgets = sorted(df["budget"].unique())
tracks = sorted(df["track"].unique())

mus_per_track = {
    "Sepang": mus,
    "YasMarina": [m for m in mus if m != 0.1],
}

def rgb_to_tikz(r, g, b):
    return f"{r:.3f},{g:.3f},{b:.3f}"

def generate_tikz_heatmap(pivot, mus_t, budgets, track,
                          cmap_name, vmin, vmax,
                          cell_w=1.0, cell_h=0.55):
    import matplotlib.cm as mcm
    import matplotlib.colors as mcolors

    cmap = mcm.get_cmap(cmap_name)
    norm = mcolors.Normalize(vmin=vmin, vmax=vmax)

    nrows = len(budgets)
    ncols = len(mus_t)

    lines = []
    lines.append(r"\resizebox{\columnwidth}{!}{%")
    lines.append(r"\begin{tikzpicture}")

    W = ncols * cell_w
    H = nrows * cell_h

    for i, b in enumerate(budgets):
        for j, mu in enumerate(mus_t):
            val = pivot.loc[b, mu] if b in pivot.index and mu in pivot.columns else np.nan
            y = nrows - 1 - i
            x0, y0 = j * cell_w, y * cell_h
            x1, y1 = x0 + cell_w, y0 + cell_h
            cx, cy = (x0 + x1) / 2, (y0 + y1) / 2

            if np.isnan(val):
                lines.append(f"  \\fill[gray!30] ({x0:.2f},{y0:.2f}) rectangle ({x1:.2f},{y1:.2f});")
                lines.append(f"  \\node[font=\\fontseries{{b}}\\selectfont\\fontsize{{5}}{{5}}\\selectfont, gray] at ({cx:.2f},{cy:.2f}) {{--}};")
            else:
                rgba = cmap(norm(val))
                rgb = rgb_to_tikz(rgba[0], rgba[1], rgba[2])
                lum = 0.299*rgba[0] + 0.587*rgba[1] + 0.114*rgba[2]
                tc = "white" if lum < 0.5 else "black"
                lines.append(f"  \\definecolor{{c{i}{j}}}{{rgb}}{{{rgb}}}")
                lines.append(f"  \\fill[c{i}{j}] ({x0:.2f},{y0:.2f}) rectangle ({x1:.2f},{y1:.2f});")
                lines.append(f"  \\node[font=\\fontseries{{b}}\\selectfont\\fontsize{{5}}{{5}}\\selectfont, {tc}] at ({cx:.2f},{cy:.2f}) {{{val:.1f}}};")

    # Grid
    for j in range(ncols + 1):
        lines.append(f"  \\draw[very thin, gray!50] ({j*cell_w:.2f},0) -- ({j*cell_w:.2f},{H:.2f});")
    for i in range(nrows + 1):
        lines.append(f"  \\draw[very thin, gray!50] (0,{i*cell_h:.2f}) -- ({W:.2f},{i*cell_h:.2f});")

    # X-axis labels (mu)
    for j, mu in enumerate(mus_t):
        lines.append(f"  \\node[below, font=\\small] at ({(j+0.5)*cell_w:.2f},0) {{{mu:.1f}}};")
    lines.append(f"  \\node[below=0.4cm, font=\\normalsize] at ({W/2:.2f},0) {{$\\mu$}};")

    # Y-axis labels (budget)
    for i, b in enumerate(budgets):
        y = nrows - 1 - i
        lines.append(f"  \\node[left, font=\\small] at (0,{(y+0.5)*cell_h:.2f}) {{{b}}};")
    lines.append(f"  \\node[left=0.5cm, font=\\normalsize, rotate=90, anchor=south] at (0,{H/2:.2f}) {{Budget (ms)}};")

    # Title
    lines.append(f"  \\node[above=0.15cm, font=\\bfseries\\normalsize] at ({W/2:.2f},{H:.2f}) {{{track}}};")

    lines.append(r"\end{tikzpicture}")
    lines.append(r"}%")
    return "\n".join(lines)


all_fastest = df_ok.groupby(["track", "mu", "budget"])["fastest_lap"].mean()
global_min = all_fastest.min()
global_max = all_fastest.max()

print(r"% Requires: \usepackage{tikz}, \usepackage{graphicx}")
print(r"\begin{figure}[t]")
print(r"\centering")

for track in tracks:
    mus_t = mus_per_track[track]
    sub = df_ok[df_ok["track"] == track].groupby(["mu", "budget"])["fastest_lap"].mean().reset_index()
    pivot = sub.pivot(index="budget", columns="mu", values="fastest_lap")
    pivot = pivot.reindex(index=budgets, columns=mus_t)
    print(generate_tikz_heatmap(pivot, mus_t, budgets, track,
                                "viridis_r", global_min, global_max))
    print(r"\vspace{0.3cm}")
    print()

print(r"\caption{CA-MCTS (Jetson Orin) fastest lap time (s) across friction coefficients and computational budgets. Darker cells indicate faster laps. Gray cells denote configurations where all trials crashed.}")
print(r"\label{fig:fastest_lap_heatmap_jetson}")
print(r"\end{figure}")
```
